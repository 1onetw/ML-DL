### Reading Data From Differrent Sources

Summary：
1. JSON by StringIO or JSON file
2. CSV file or HTML including csv data
3. HTML including table by match key word
4. excel
5. pickle

```python
import pandas as pd
from io import StringIO

## Data已经是json字符串，所以不需要再进行json解析
Data = '{"employee_name": "James", "email": "james@gmail.com", "job_profile": [{"title1":"Team Lead", "title2":"Sr. Developer"}]}'
## orient 默认为column, 所以列名的值至少有一个需要是列表
df=pd.read_json(StringIO(Data), orient="column")
## Pandas 推荐统一使用类文件对象（如 StringIO）或路径字符串来处理所有 I/O 操作，保持 API 一致性
df1=pd.read_json(Data, orient="column")
```

```python
"""
## json、pickle、StringIO区别
1. json:
    - 支持多种语言
    - 序列化格式：文本(字符串)
    - 扩展名：.json

2. pickle
    - 只支持Python
    - 序列化格式：二进制
    - 扩展名：.pkl

3. StringIO
    - 序列化格式：文件流(类文件对象)，需通过.read()方法读取
"""
```
