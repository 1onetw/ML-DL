### Assignment 3: Array Operations

```python
array1=np.random.randint(1,21,size=(3,4))
array2=np.random.randint(1,21,size=(3,4))

print("Array1:\n", array1)
print("Array2:\n", array2)

array_wise=array1 + array2
print("array_wise:\n", array_wise)

array_sub=array1 - array2
print("array_sub:\n", array_sub)

array_mul=array1 * array2
print("array_mul:\n", array_mul)

array_div=array1 / array2
print("array_div:\n", array_div)
```

### Assignment 4: Statistical Operations

1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.

```python
array=np.random.randint(1,21,size=(5,5))
print("Array:\n",array)

mean=np.mean(array)
print("Mean:\n",mean)

median=np.median(array)
print("Median:\n",median)

std=np.std(array)
print("Standard Deviation:\n",std)

var=np.var(array)
print("Variance:\n",var)
```

2. Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

```python
array=np.arange(1,10).reshape(3,3)
print("Array:\n", array)

mean=np.mean(array)
std=np.std(array)

normalized_array=(array-mean)/std
print("Normalized array:\n", normalized_array)
```

### Assignment 5: Broadcasting

1. Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.

```python
array=np.random.randint(1,21,size=(3,3))
row_array=np.random.randint(1,21,size=(3,))
print("Original array:\n", array)
print("Row array:\n", row_array)

# 利用广播机制对二维数组逐行加上一维数组
result = array + row_array
print("Result:\n", result)
```

2. Create a NumPy array of shape (4, 4) filled with random integers. Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

```python
array=np.random.randint(1,21,size=(4,4))
column_array=np.random.randint(1,21,size=(4,))
print("Original array:\n", array)
print("Row array:\n", column_array)

print("Column array:\n", column_array[:, np.newaxis])
# 利用广播机制对二维数组逐列减去一维数组
result = array - column_array[:, np.newaxis]
print("Result:\n", result)
```

### Assignment 6: Linear Algebra

1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.

```python
matrix=np.random.randint(1,21,size=(3,3))
print("Matrix:\n", matrix)

# Compute the determinant
determinant=np.linalg.det(matrix)
print("Determinant:\n", determinant)

# Compute the inverse
inverse=np.linalg.inv(matrix)
print("Inverse:\n", inverse)

# Compute the eigenvalues and eigenvectors
eigenvalues, eigenvectors=np.linalg.eig(matrix)
print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

### Assignment 9: Structured Arrays

1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

```python
# Create a structured array with fields 'name' and 'age' and 'weight'
data_type = [('name','U10'), ('age','i4'), ('weigth','f4')]
data = np.array([('Alice',25,55.5), ('Bob',30,70.0), ('Charlie',35,80.0)], dtype=data_type)
print("Original array:\n", data)

# Sort the array by 'age'
sorted_data = np.sort(data, order='age')
print("Sorted array:\n", sorted_data)
```

2. Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.

```python
# Create a structured array with fields 'x' and 'y'
data_type = [('x','i4'),('y','i4')]
data = np.array([(1,2),(3,4),(5,6)], dtype=data_type)
print("Original array:\n", data)

# Compute the Euclidean distance between each pair of points
## NumPy会自动广播，生成一个n×n的矩阵，其中每个元素(i,j)表示第i个点与第j个点的x坐标差值
distances = np.sqrt((data['x'][:, np.newaxis] - data['x'])**2 + (data['y'][:, np.newaxis] - data['y'])**2)
print("Euclidean distances:\n", distances)
```

### Assignment 10: Masked Arrays(掩码数组)

1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.

```python
import numpy.ma as ma
# Create a masked array of shape(4,4) with random integers
array=np.random.randint(1,21,size=(4,4))
masked_array=ma.masked_array(array, mask=array>10)
print("Original array:\n", array)
print("Masked array:\n", masked_array)

# Compute the sum of the unmasked elements
sum_unmasked=ma.sum(masked_array)
print("Sum of unmasked elements:\n", sum_unmasked)
```

2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
# Create a masked array of shape(3,3) with random integers
array=np.random.randint(1,21,size=(3,3))
masked_array=ma.masked_array(array, mask=np.eye(3,dtype=bool))
print("Original array:\n", array)
print("Masked array:\n", masked_array)

# Replace the masked elements with the mean of the unmasked elements
mean_unmasked=ma.mean(masked_array)
masked_array=ma.filled(masked_array, mean_unmasked)
print("Masked array with mean filled:\n", masked_array)
```
