### Data Visualization With Seaborn
Seaborn is a Python visualization library based on Matplotlib that provides a high-level inerface for drawing attractive and informative statistical graphics.Seaborn helps in creating complex visualizations with just a few lines of code.

Summary:
1. scatter plot
2. line plot
3. bar plot
4. box plot
5. violin plot
6. histplot
7. kdeplot
8. pairplot
9. heatmap
10. A application of sales data visualization with seaborn and matplotlib

```python
import seaborn as sns
```

```python
### Basic Plotting With Seaborn
tips=sns.load_dataset("tips")
tips
```

```python
## create a scatter plot
import matplotlib.pyplot as plt

### with seaborn better
sns.scatterplot(x="total_bill",y='tip',data=tips)
plt.title("Scatter plot of total bill vs tip")
plt.show()

### with matplotlib
plt.scatter(tips['total_bill'],tips['tip'])
plt.title("Scatter plot of total bill vs tip")
plt.show()
```

```python
## Line plot

### with seaborn better
sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line plot of Total Bill by size")
plt.show()

### with matplotlib
plt.plot(tips['size'],tips['total_bill'])
plt.title("Line plot of Total Bill by size")
plt.show()
```

```python
## bar plot

### with seaborn better
# Seaborn 自动计算各 day 的 total_bill 平均值
sns.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar Plot of Total Bill by Day")
plt.show()

### with matplotlib
# Matplotlib 直接绘制所有 total_bill 值（可能重叠），若需计算平均值需手动聚合再绘制
plt.bar(tips['day'],tips['total_bill'])
plt.title("Bar Plot of Total Bill by Day")
plt.show()
```

```python
## Box Plot
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
## Violin Plot

sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
## Histogram

### seaborn better
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
### KDE Plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
## Pair Plot 成对关系图,也称为 散点图矩阵, 针对于数值变量
sns.pairplot(tips, kind='reg', diag_kind='kde')
sns.pairplot(tips,hue='sex', kind='reg')
sns.pairplot(tips,hue='smoker')
```

```python
## heatmap 热力图，针对于数值变量
### correlation：相关性
corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True)
```

```python
## sales data visualization with seaborn and matplotlib
import pandas as pd
sales_df=pd.read_csv("sales_data.csv")
sales_df.head()
```

```python
sales_df.info()
```

```python
## bar plot total sales by product
plt.figure(figsize=(10,6))
### estimator：估计器，用于放聚合函数，默认是均值
sns.barplot(x='Product Category',y="Total Revenue",data=sales_df,estimator=sum)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
## bar plot total sales by Region
plt.figure(figsize=(10,6))
### estimator：估计器，用于放聚合函数，默认是均值
sns.barplot(x='Region',y="Total Revenue",data=sales_df,estimator=sum)
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```
